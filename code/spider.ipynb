```python
import demjson3 as demjson
import sys
sys.modules['demjson'] = demjson

from Crypto.Cipher import AES

import gopup as gp
import pandas as pd
import matplotlib.pyplot as plt
# 在运行时设置了虚拟环境
```

```python
target_year = 2016  # 替换查询的年份
cookie = ''

keywords = ["洪荒之力","友谊的小船","定个小目标","吃瓜群众","葛优躺","辣眼睛","全是套路","蓝瘦香菇","老司机","厉害了我的哥"]  # 替换查询的关键词列表

start_date = '2015-01-01'
end_date = '2018-01-01'

all_data = pd.DataFrame()
final_data = pd.DataFrame()
```

```python
# 循环抓取每个关键词
for word in keywords:
    print(f"正在抓取：{word}")
    try:
        # 调用gopup获取百度指数
        df = gp.baidu_search_index(word=word, start_date=start_date, end_date=end_date, cookie=cookie)
        if df.empty:
            print(f"{word} 无数据，跳过")
            continue

        # 重置索引，提取日期列
        df = df.reset_index().rename(columns={"index": "日期"})
        # 统一列名
        df.rename(columns={"all": "全部", "pc": "PC端", "wise": "移动端"}, inplace=True)
        # 添加关键词、年份、地区列
        df["关键词"] = word
        df["年份"] = target_year
        df["地区"] = "全国"

        
        df = df[["年份", "关键词", "地区", "日期", "全部", "PC端", "移动端"]]

        # 合并数据
        final_data = pd.concat([final_data, df], ignore_index=True)

    except Exception as e:
        print(f"{word} 抓取失败：{str(e)}")
```

```python
# 保存为CSV
output_file = f"baidu_index_multiple_keywords2016.csv"
final_data.to_csv(output_file, index=False, encoding="utf-8-sig")

print(f"\n生成完成：{output_file}")
```
